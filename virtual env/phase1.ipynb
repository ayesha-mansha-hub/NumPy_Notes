## Numpy Array and basics


```python
import numpy as np
```

### List VS NumPy Array

```python
import time
st = time.time()
l = [i*i for i in range(1000000)]
print("List Time: " , time.time() - st)

start = time.time()
array = np.arange(1000000, dtype= np.int32)
print("NumPy array Time: " , time.time() - start)
```

### Creating NumPy Arrays

```python
arr = np.array([[2,3,4,5],[6,7,8,3]])
print(arr)
```

```python
# zeros array
zero = np.zeros((3,4))
print(zero)
```

```python
# Ones array
one = np.ones((4,5))
print(one)
```

```python
# creating an array with another array with constant number as its all entries
full = np.full((3,4),7,dtype=np.int32)
print(full)
```

```python
#creating array with random values 
random = np.random.random((4,5))  #generates random float numbers between 0.0 and 1.0
print(random)
```

```python
# sequence 
arr = np.arange(0,100,2, dtype=np.int32)
print("Sequence Array: \n", arr)
```

### Vector, Matrix and tensor

```python
# 1D -> vector 
vector = np.array([2,3])
print(vector)
```

```python
# 2D -> Matrix
matrix = np.array([[3,4,5],
                   [7,8,4]])
print(matrix)
```

```python
# 3D and 3D+ -> Tensor
Tensor = np.array([[[3,4,5],[5,8,0]],
                    [[0,9,1],[1,2,7]]])
print(Tensor)
```

## Array Properties

```python
# getting shape
arr = np.array([[2,3,4],
                 [5,6,7]])
print("Shape of Tensor: ", arr.shape)  #In general, all (1D, 2D and so on) are called Tensor
# getting Dimensions
print("Dimensions of Tensor: ", arr.ndim)
# getting size -> It tells total number of elements
print("Size of Tensor: ", arr.size)
# getting dtype (datatype)
print("Datatype of Tensor: ", arr.dtype)
# getting size per element in bytes_ here it is 8bytes bcz my PC is 64bit
print("Size per element in bytes: ", arr.itemsize)
# getting total size in bytes
print("Total size in Bytes of Tensor: ", arr.nbytes) # size (here it is 6) * itemsize (8 bytes per element) = 48 bytes 
```

### Reshaping Array

```python
org = np.arange(12)
print("Original Array: ", org)

reshaped = org.reshape((3,4))
print("Reshaped Array: \n", reshaped)

# Flattening array
flattened = reshaped.flatten()
print("Flattened Array: ", flattened)
```

### Ravel()
##### It returns a "view", instead of copy
##### It also takes a multi-dimensional array and stretches it out into a single line of numbers

```python
raveled = reshaped.ravel()
print("Raveled array: ", raveled)
```

### Transpose

```python
transpose = reshaped.T  #coverting rows into columns and columns into rows
print("transposed Matrix: \n", transpose)
```
